```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

data = pd.read_csv("electricity.csv")

sns.lineplot(
    x="Date",
    y="Electricity Consumption",
    hue="Consumer Category",
    linestyle="--",
    style="Consumer Category",
    marker="o",
    data=data
)

plt.xlabel("Time lines")
plt.ylabel("Electricity Consumption (kWh)")
plt.title("Monthly Electricity Consumption Trend")

plt.xticks(
    rotation=45,
    ha="center",
    fontsize=12,
    fontfamily="serif",
    fontweight="bold"
)

plt.yticks(
    rotation=45,
    ha="center",
    fontsize=12,
    fontfamily="serif",
    fontweight="bold"
)

plt.grid(True)
plt.show()
```
